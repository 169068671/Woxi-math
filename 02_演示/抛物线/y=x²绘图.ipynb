# 用 Woxi 绘制 $y=x^2$

这是一个使用本地 **JupyterLab + Woxi (Wolfram Language)** 内核绘制抛物线的最小示例。

## 1. 检查 Woxi 内核

运行后应输出 `3`。

```wolfram
1 + 2
```

## 2. 绘制函数

下面在区间 $-5\le x\le5$ 上绘制 $y=x^2$。

```wolfram
Plot[
 x^2,
 {x, -5, 5},
 AxesLabel -> {"x", "y"},
 PlotLabel -> "y = x^2",
 PlotStyle -> Blue,
 GridLines -> Automatic,
 ImageSize -> 600
]
```

## 3. 检查结果

图像应为开口向上的抛物线，顶点在 $(0,0)$，并关于 $y$ 轴对称。
